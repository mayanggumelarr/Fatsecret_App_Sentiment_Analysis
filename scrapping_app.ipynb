```python
import pandas as pd
import numpy as np
from google_play_scraper import app
from google_play_scraper import Sort, reviews
from nltk.sem.chat80 import country
```

```python
result, continuation_token = reviews(
    # https://play.google.com/store/apps/details?id=com.fatsecret.android
    'com.fatsecret.android',
    lang='id',
    country='id',
    sort=Sort.NEWEST,
    count=10500,
    filter_score_with=None
)
```

```python
df_scraped = pd.DataFrame(np.array(result), columns=['reviews'])
df_scraped = df_scraped.join(pd.DataFrame(df_scraped.pop('reviews').tolist()))
df_scraped.head()
```
